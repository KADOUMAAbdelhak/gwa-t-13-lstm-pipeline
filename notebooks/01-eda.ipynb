# Exploratory Data Analysis (EDA) for GWA-T-13 Materna Dataset

In this notebook, we will perform exploratory data analysis on the GWA-T-13 Materna dataset. We will load the data, visualize it, and analyze its characteristics to understand its structure and the relationships between variables.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set(style='whitegrid')
```

```python
# Load the dataset
data_path = '../data/raw/GWA-T-13/'  # Adjust the path as necessary
files = ['file1.csv', 'file2.csv']  # Replace with actual file names

dataframes = []
for file in files:
    df = pd.read_csv(data_path + file)
    dataframes.append(df)

# Combine all dataframes into one
data = pd.concat(dataframes, ignore_index=True)
data.head()
```

```python
# Check for missing values
missing_values = data.isnull().sum()
missing_values[missing_values > 0]
```

```python
# Visualize the distribution of key variables
plt.figure(figsize=(12, 6))
sns.histplot(data['key_variable'], bins=30, kde=True)
plt.title('Distribution of Key Variable')
plt.xlabel('Key Variable')
plt.ylabel('Frequency')
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(10, 8))
correlation_matrix = data.corr()
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', square=True)
plt.title('Correlation Heatmap')
plt.show()
```

```python
# Time series visualization
plt.figure(figsize=(14, 7))
plt.plot(data['timestamp'], data['key_variable'], label='Key Variable')
plt.title('Time Series of Key Variable')
plt.xlabel('Time')
plt.ylabel('Key Variable')
plt.legend()
plt.show()
```
