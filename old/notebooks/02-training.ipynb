# Training Models on GWA-T-13 Materna Dataset

In this notebook, we will train LSTM models on the GWA-T-13 Materna dataset. We will load the data, preprocess it, create sliding windows, and then train the models. Finally, we will evaluate the models and visualize the results.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.data.load_data import load_data
from src.data.preprocessing import preprocess_data
from src.data.windowing import create_sliding_windows
from src.models.lstm_model import LSTMModel
from src.training.train import train_model
from src.evaluation.evaluate import evaluate_model
from src.utils.config import load_config

# Load configuration
config = load_config('configs/config.yaml')
```

```python
# Load the dataset
data = load_data(config['data']['raw_data_path'])
data.head()
```

```python
# Preprocess the data
processed_data = preprocess_data(data)
processed_data.head()
```

```python
# Create sliding windows
X, y = create_sliding_windows(processed_data, config['data']['window_size'])
X.shape, y.shape
```

```python
# Initialize and train the LSTM model
lstm_model = LSTMModel(config['model'])
history = train_model(lstm_model, X, y, config['training'])
```

```python
# Evaluate the model
evaluation_results = evaluate_model(lstm_model, X, y)
evaluation_results
```

```python
# Visualize training history
plt.figure(figsize=(12, 6))
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(loc='upper right')
plt.savefig('outputs/figures/training_loss.png')
plt.show()
```

```python
# Save evaluation results
evaluation_results.to_csv('outputs/results/evaluation_results.csv', index=False)
```
